```python
import pandas as pd

url = "https://raw.githubusercontent.com/mizgan/Resume-Screening/master/UpdatedResumeDataSet.csv"
df = pd.read_csv(url)
df.head(4)
```

```python
rows, columns = df.shape  
print(f"The dataset contains {rows} rows and {columns} columns.")
df.shape
```

```python
df.dtypes
```

```python
missing_values = df.isnull().sum()
print(missing_values)

if missing_values.sum() == 0:
    print("\nthere are no missing values in any column.")
else:
    print(f"\nThere are {missing_values.sum()} missing values in total.")
```
